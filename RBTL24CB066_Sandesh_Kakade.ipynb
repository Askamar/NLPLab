```python
import re

# Sample Text
text = """
Hello, my name is Sandesh.

My email is sandeshkakade35@gmail.com and my office email is
sandesh@company.co.in.

My phone numbers are 8799948909 and +91-9922994701.

Meeting Date: 15/08/2026
Joining Date: 06/08/2025

I live in Pune, Maharashtra.
I study at JSPM Rajarshi Shahu College of Engineering.
"""

# 1. Email Extraction
email_pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b'

emails = re.findall(email_pattern, text)

print("===== EMAIL IDs =====")
for email in emails:
    print(email)


# 2. Phone Number Extraction
phone_pattern = r'(?:\+91[- ]?)?[6-9]\d{9}'

phones = re.findall(phone_pattern, text)

print("\n===== PHONE NUMBERS =====")
for phone in phones:
    print(phone)

# 3. Date Extraction
date_pattern = r'\b(?:0[1-9]|[12][0-9]|3[01])/(?:0[1-9]|1[0-2])/\d{4}\b'

dates = re.findall(date_pattern, text)

print("\n===== DATES =====")
for date in dates:
    print(date)


# 4. Simple Named Entity Recognition
# Finds words starting with capital letters

entity_pattern = r'\b[A-Z][a-z]+(?:\s+[A-Z][a-z]+)*\b'

entities = re.findall(entity_pattern, text)

print("\n===== NAMED ENTITIES =====")
for entity in entities:
    print(entity)


# 5. Finite State Automaton (FSA)
# Recognizes the word "om"


def finite_state_automaton(word):

    state = 0

    for ch in word:

        if state == 0:
            if ch == 'o':
                state = 1
            else:
                return False

        elif state == 1:
            if ch == 'm':
                state = 2
            else:
                return False

        elif state == 2:
            return False

    return state == 2


print("\n===== FINITE STATE AUTOMATON =====")

words = ["om", "on", "of", "ram", "oms", "sam"]

for word in words:
    if finite_state_automaton(word):
        print(word, "-> ACCEPTED")
    else:
        print(word, "-> REJECTED")



# 6. Finite State Automaton (FSA)
# Recognizes the word "sandesh"

def finite_state_automaton(word):
    state = 0

    for ch in word:
        if state == 0:
            if ch == 's':
                state = 1
            else:
                return False

        elif state == 1:
            if ch == 'a':
                state = 2
            else:
                return False

        elif state == 2:
            if ch == 'n':
                state = 3
            else:
                return False

        elif state == 3:
            if ch == 'd':
                state = 4
            else:
                return False

        elif state == 4:
            if ch == 'e':
                state = 5
            else:
                return False

        elif state == 5:
            if ch == 's':
                state = 6
            else:
                return False

        elif state == 6:
            if ch == 'h':
                state = 7
            else:
                return False

    return state == 7

print("\n===== FINITE STATE AUTOMATON =====")

words = ["sandesh", "sande", "sand", "ram", "sandess", "om"]

for word in words:
    if finite_state_automaton(word):
        print(word, "-> ACCEPTED")
    else:
        print(word, "-> REJECTED")
```
